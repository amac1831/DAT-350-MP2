```python
# import libraries

import pandas as pd
import matplotlib.pyplot as plt
# import libraries
```

```python
rawDF = pd.read_csv("mlbbatting1901-2021.csv")

rawDF.tail()
```

```python
rawDF.describe()
```

```python
def printAmtNA(aDF):
   for aCol in aDF:
    print(f"The amount of missing values in {aCol} is {aDF[aCol].isna().sum()}")

printAmtNA(rawDF)
```

```python
rawDF["Date"].dtype

rawDF["Date"] = pd.to_datetime(rawDF["Date"], errors="coerce")

rawDF["Date"].dtype

rawDF["Year"] = rawDF["Date"].dt.year

rawDF[["Date", "Year"]].head()
```

```python
season_df = rawDF.groupby(['Year', 'Tm']).agg({
    'AB': 'sum',
    'H': 'sum',
    'HR': 'sum',
    'R': 'sum',
    'RBI': 'sum',
    'BB': 'sum',
    'SO': 'sum',
    'PA': 'sum'
}).reset_index()

season_df['AVG'] = season_df['H'] / season_df['AB']

season_stats = season_df[season_df["AB"] > 200]

season_stats.shape

season_stats.head()
```

```python
plt.figure(figsize=(8,5))
plt.hist(season_stats["AVG"])
plt.xlabel("Team Batting Average")
plt.ylabel("Number of Team-Seasons")
plt.title("Distribution of Team Batting Averages (1901-2021)")
plt.figtext(0.5, 0.01, "Fig 1: Distribution of team batting averages across MLB seasons from 1901–2021, showing most clustered around .240–.280.", wrap=True, horizontalalignment='center', fontsize=10)
plt.subplots_adjust(bottom=0.2)
plt.show()
```

**Fig 1:** Distribution of team batting averages across MLB seasons from 1901–2021, showing most clustered around .240–.280.

```python
league_avg = season_stats.groupby("Year")["AVG"].mean()

plt.figure(figsize=(8,5))
plt.plot(league_avg.index, league_avg.values)
plt.xlabel("Year")
plt.ylabel("League Average Batting Average")
plt.title("League Batting Average Over Time")
plt.figtext(0.5, 0.01, "Fig 2: League-wide average team batting average from 1901–2021, showing fluctuations across baseball eras and a general decline in modern times.", wrap=True, horizontalalignment='center', fontsize=10)
plt.subplots_adjust(bottom=0.2)
plt.show()
```

**Fig 2:** League-wide average team batting average from 1901–2021, showing fluctuations across baseball eras and a general decline in modern times.

```python
hr_trend = season_stats.groupby("Year")["HR"].mean()

plt.figure(figsize=(8,5))
plt.plot(hr_trend.index, hr_trend.values, color = 'green')
plt.xlabel("Year")
plt.ylabel("Average Team Home Runs")
plt.title("Average Team Home Runs Over Time")
plt.figtext(0.5, 0.01, "Fig 3: Average team home runs per season from 1901–2021, showing a dramatic increase in power hitting over time.", wrap=True, horizontalalignment='center', fontsize=10)
plt.subplots_adjust(bottom=0.2)
plt.show()
```

**Fig 3:** Average team home runs per season from 1901–2021, showing a dramatic increase in power hitting over time.

```python
so_trend = season_stats.groupby("Year")["SO"].mean()

plt.figure(figsize=(8,5))
plt.plot(so_trend.index, so_trend.values, color = 'red')
plt.xlabel("Year")
plt.ylabel("Average Team Strikeouts")
plt.title("Average Team Strikeouts Over Time")
plt.figtext(0.5, 0.01, "Fig 4: Average team strikeouts per season from 1901–2021, showing a steady increase and highlighting the modern rise in strikeouts.", wrap=True, horizontalalignment='center', fontsize=10)
plt.subplots_adjust(bottom=0.2)
plt.show()
```

**Fig 4:** Average team strikeouts per season from 1901–2021, showing a steady increase and highlighting the modern rise in strikeouts.

```python
plt.figure(figsize=(8,5))
plt.scatter(season_stats["SO"], season_stats["AVG"], color = 'purple')
plt.xlabel("Team Strikeouts")
plt.ylabel("Team Batting Average")
plt.title("Strikeouts vs Team Batting Average")
plt.figtext(0.5, 0.01, "Fig 5: Relationship between team strikeouts and team batting average, showing that teams with higher strikeouts tend to have slightly lower batting averages.", wrap=True, horizontalalignment='center', fontsize=10)
plt.subplots_adjust(bottom=0.2)
plt.show()
```

**Fig 5:** Relationship between team strikeouts and team batting average, showing that teams with higher strikeouts tend to have slightly lower batting averages.
